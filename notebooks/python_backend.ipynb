```python
from flask import Blueprint, request, jsonify
import os

prescription_bp = Blueprint("prescription", __name__)

ALLOWED_EXTENSIONS = {"jpg", "jpeg", "png"}


def allowed_file(filename):
    return (
        "." in filename
        and filename.rsplit(".", 1)[1].lower() in ALLOWED_EXTENSIONS
    )


@prescription_bp.route("/predict", methods=["POST"])
def predict():
    if "image" not in request.files:
        return jsonify({
            "success": False,
            "message": "No image uploaded"
        }), 400

    image = request.files["image"]

    if image.filename == "":
        return jsonify({
            "success": False,
            "message": "No image selected"
        }), 400

    if not allowed_file(image.filename):
        return jsonify({
            "success": False,
            "message": "Invalid file type. Only JPG, JPEG and PNG are allowed"
        }), 400

    try:
        upload_folder = os.path.join(
            os.path.dirname(os.path.dirname(__file__)),
            "uploads"
        )

        os.makedirs(upload_folder, exist_ok=True)

        file_path = os.path.join(upload_folder, image.filename)
        image.save(file_path)

        return jsonify({
            "success": True,
            "message": "Image received successfully",
            "filename": image.filename
        }), 200

    except Exception:
        return jsonify({
            "success": False,
            "message": "File upload failed"
        }), 500
```

```python
def validate_input(data):
    # Check if data exists
    if not data:
        return False

    # Check if data is a dictionary
    if not isinstance(data, dict):
        return False

    # Check that at least one value is provided
    valid_value = False

    for value in data.values():
        if value is not None and str(value).strip() != "":
            valid_value = True
            break

    if not valid_value:
        return False

    return True
```

```python
import os
import uuid

from flask import Blueprint, request, jsonify
from werkzeug.utils import secure_filename


prescription_bp = Blueprint("prescription", __name__)

UPLOAD_FOLDER = "uploads"

os.makedirs(UPLOAD_FOLDER, exist_ok=True)


# Allowed file types
ALLOWED_EXTENSIONS = {"jpg", "jpeg", "png", "pdf"}


def allowed_file(filename):
    return (
        "." in filename
        and filename.rsplit(".", 1)[1].lower() in ALLOWED_EXTENSIONS
    )


@prescription_bp.route("/upload", methods=["POST"])
def upload_prescription():

    # Check whether file was uploaded
    if "file" not in request.files:
        return jsonify({
            "success": False,
            "error": "No file uploaded"
        }), 400

    file = request.files["file"]

    # Check whether a file was selected
    if file.filename == "":
        return jsonify({
            "success": False,
            "error": "No file selected"
        }), 400

    # Validate file type
    if not allowed_file(file.filename):
        return jsonify({
            "success": False,
            "error": "Invalid file type. Only JPG, JPEG, PNG and PDF files are allowed."
        }), 400

    # Secure the original filename
    original_filename = secure_filename(file.filename)

    # Get file extension
    extension = original_filename.rsplit(".", 1)[1].lower()

    # Generate a unique filename
    filename = f"{uuid.uuid4().hex}.{extension}"

    # Create complete file path
    file_path = os.path.join(UPLOAD_FOLDER, filename)

    # Save file
    file.save(file_path)

    # Send success response
    return jsonify({
        "success": True,
        "message": "Prescription uploaded successfully",
        "filename": filename
    }), 200
```

```python
database_file = "/content/drive/MyDrive/Mediscan_project/Python_Backend/database/database.py"

code = '''
import os
import sqlite3

BASE_DIR = os.path.dirname(os.path.abspath(__file__))

DATABASE_PATH = os.path.join(BASE_DIR, "mediscan.db")


def get_db_connection():
    """
    Create and return a connection to the SQLite database.
    """
    connection = sqlite3.connect(DATABASE_PATH)

    connection.row_factory = sqlite3.Row

    return connection


def init_db():
    """
    Create the prescriptions table if it does not already exist.
    """

    connection = get_db_connection()

    cursor = connection.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS prescriptions (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            filename TEXT NOT NULL,
            file_path TEXT NOT NULL,
            uploaded_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
            status TEXT DEFAULT 'uploaded'
        )
    """)

    connection.commit()
    connection.close()


if __name__ == "__main__":
    init_db()
    print("✅ Mediscan SQLite database initialized successfully!")
'''

with open(database_file, "w") as file:
    file.write(code)

print("✅ database.py created successfully!")
print("📁", database_file)
```
